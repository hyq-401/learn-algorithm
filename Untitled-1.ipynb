```python
from collections import Counter


class Solution:
    def isAnagram(self, s: str, t: str) -> bool:
        # 长度不同直接短路，避免无谓的两次 O(n) 计数
        if len(s) != len(t):
            return False
        return Counter(s) == Counter(t)
```

```python
class Solution(object):
    def isAnagram(self, s: str, t: str) -> bool:
        from collections import Counter

        a_count = Counter(s)
        b_count = Counter(t)
        return a_count == b_count
```
